```python
# imports

import os, sys, requests, json, time
from datetime import datetime
```

```python
current_year = str(datetime.now().year)

def fetch_earnings_call_transcript(api_key, symbol="AAPL", quarter="Q1", year=current_year):
    qtr = f"{year}{quarter}"
    url = f'https://www.alphavantage.co/query?function=EARNINGS_CALL_TRANSCRIPT&symbol={symbol}&quarter={qtr}&apikey={api_key}'
    r = requests.get(url)
    data = r.json()
    return data
```
